Exercício 3 - Funções + Lista + Condicional

```python
def verificar_presentes(lista_presentes):
  quantidade = len(lista_presentes)
  if quantidade >= 3:
    print("O Papai Noel terá trabalho este ano!")
  else:
    print("Sua lista está muito curta!")

  return quantidade

lista = ["Caderno", "Computador", "Fones de ouvido", "Um gatinho"]
total = verificar_presentes(lista)
print(f"Resultado do retorno: A lista tem {total} presentes.")

```

Exercício 4 - Função + Dicionário + Laços

```python
def contar_enfeites(enfeites):
  total_enfeites = 0
  for quantidade in enfeites.values():
    total_enfeites += quantidade
  return total_enfeites

enfeite = {"Laço": 8, "Bolinhas azuis": 12, "Bolinhas roxas": 12, "Estrela": 1, "Bengala doce": 6}

print('--- Enfeites na árvore ---')
for itens in enfeite:
    print(itens)

# Chama a função para contar o total
total = contar_enfeites(enfeite)

print("\n--- Resultado ---")
print(f"O total de enfeites na sua árvore é: {total}")
```

Exercício 5 - Função completa

```python
ceia_de_natal = {}

def resumo_ceia(dados):
  num_pessoas = len(dados)

  total_pratos = 0

  for lista_pratos in dados.values():
    total_pratos += len(lista_pratos)

  print("\n" + "="*40)
  print(" RESUMO DA CEIA DE NATAL 🎄")
  print("="*40)

  print(f"Número de pessoas contribuindo: {num_pessoas}")
  print(f"Total de pratos levados: {total_pratos}")

  if total_pratos > 10:
    print("\n Essa ceia vai ser farta!")

  print("="*40)



print("🌟 BEM-VINDO(A) AO ORGANIZADOR DE CEIA DE NATAL 🌟")
print("Digite 'sair' a qualquer momento no nome da pessoa para terminar o cadastro.")

while True:
  nome = input("\nNome da pessoa que vai contribuir (ou 'sair'): ").strip()

  if nome.lower() == 'sair':
    break

  pratos_da_pessoa = []
  print(f"\n--- Cadastro dos pratos de {nome} ---")

  while True:
    prato = input(f"Prato que {nome} levará (ou 'ok' para finalizar os pratos dela): ").strip()

    if prato.lower() == 'ok':
      if pratos_da_pessoa:
        break
      else:
        print(" Você deve cadastrar pelo menos um prato.")
    elif prato:
      pratos_da_pessoa.append(prato)

  if pratos_da_pessoa:
    ceia_de_natal[nome] = pratos_da_pessoa


if ceia_de_natal:
  resumo_ceia(ceia_de_natal)
else:
  print("\nNenhuma contribuição foi cadastrada. A ceia está vazia!")
```
